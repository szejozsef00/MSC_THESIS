### Basic

```python
import os, pickle, gc, torch
import pandas as pd, numpy as np
from neuralforecast import NeuralForecast
from neuralforecast.models import NHITS

gc.collect()
torch.cuda.empty_cache()
os.environ["PYTORCH_CUDA_ALLOC_CONF"] = "max_split_size_mb:128"

INPUT_PATH      = "/dbfs/mnt/thesis/output_data/"
CHECKPOINT_PATH = "/dbfs/mnt/thesis/models/nhits/basic/nhits_tuning/"
BEST_CFG_PATH   = os.path.join(CHECKPOINT_PATH, "best_cfg_2010-08-08.pkl")

START_DATE  = pd.Timestamp("2010-08-01 00:00:00")
H = 288
VAL_DAYS = 7
VAL_SIZE = VAL_DAYS * H

df = pd.read_csv(INPUT_PATH + "processed_data.csv", parse_dates=["DATETIME"])
df = df.rename(columns={"DATETIME":"ds","LOCATION":"unique_id","VALUE":"y"})
df = df[df["unique_id"].isin(df["unique_id"].unique()[:48])]

history = df[df["ds"] < START_DATE].sort_values(["unique_id","ds"])

best_cfg = pickle.load(open(BEST_CFG_PATH, "rb"))
best_cfg["mlp_units"] = [list(u) for u in best_cfg["mlp_units"]]
```

```python
model = NHITS(
    **best_cfg,
    precision="16-mixed",
    max_steps=1000,
    val_check_steps=50,
    early_stop_patience_steps=5,
    enable_checkpointing=True,
)

nf = NeuralForecast(models=[model], freq="5min")
print("⏳ Retraining on data up to", START_DATE.date())
nf.fit(df=history, val_size=VAL_SIZE)

ids   = history["unique_id"].unique()
dates = pd.date_range(start=START_DATE, periods=H, freq="5min")
future = pd.DataFrame({
    "unique_id": np.repeat(ids, len(dates)),
    "ds":         np.tile(dates, len(ids))
})
Y_hat = nf.predict(df=history, futr_df=future)
Y_hat.rename(columns={"unique_id":"LOCATION","ds":"DATETIME"}, inplace=True)

Y_hat.to_csv(f"/dbfs/mnt/thesis/predictions/nhits/basic_weekly_retune/predictions_{START_DATE.date()}.csv", index=False)
```
