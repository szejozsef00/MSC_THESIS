```python
import os
import gc
import pickle
import torch
import pandas as pd
import optuna
import plotly.graph_objects as go
import datetime
import numpy as np
from neuralforecast import NeuralForecast
from neuralforecast.models import NHITS
from neuralforecast.auto import AutoNHITS
from neuralforecast.losses.pytorch import MAE
from sklearn.metrics import mean_absolute_error

os.environ["PYTORCH_CUDA_ALLOC_CONF"] = "max_split_size_mb:128"
INPUT_PATH   = "/dbfs/mnt/thesis/output_data/"
SAVE_PATH    = "/dbfs/mnt/thesis/models/nhits/basic/nhits_tuning/"
FORECAST_DAY = pd.Timestamp("2010-08-01 00:00:00")
os.makedirs(SAVE_PATH, exist_ok=True)

H = 288
VAL_DAYS = 7
VAL_SIZE = VAL_DAYS * H
```

```python
# read processed data
df_all = pd.read_csv(
    os.path.join(INPUT_PATH, "processed_data.csv"),
    parse_dates=["DATETIME"]
)
df_all = df_all.rename(
    columns={"DATETIME": "ds", "LOCATION": "unique_id", "VALUE": "y"}
)
df_train = (
    df_all[df_all["ds"] < FORECAST_DAY]
    .sort_values(["unique_id", "ds"])
)
data = df_train[df_train["unique_id"].isin(df_train["unique_id"].unique()[:48])]
```

```python
import subprocess
out = subprocess.check_output([
    "nvidia-smi",
    "--query-compute-apps=pid,process_name,used_gpu_memory",
    "--format=csv,noheader,nounits"
])
print(out.decode())
```

```python
def nhits_search_space(trial):
    return {
        "input_size": trial.suggest_categorical("input_size", [2*H]),
        "learning_rate": trial.suggest_float("learning_rate", 1e-4, 1e-2, log=True),
        "batch_size": 8,
        "windows_batch_size": trial.suggest_categorical("windows_batch_size", [32]),
        "n_blocks" : trial.suggest_categorical("n_blocks", [
            [1,1,1,1],
            [2,2,2,2],
        ]),
        "mlp_units": trial.suggest_categorical("mlp_units", [
            [[256, 256], [256, 256], [256, 256], [256, 256]],
            [[512, 512], [512, 512], [512, 512], [512, 512]],
        ]),

        "n_freq_downsample": trial.suggest_categorical("n_freq_downsample", [
            [144,  36,  12,  1],
            [144,  12,  6,  1],
            [72,  12,  3,  1],
            [12,  6,  3,  1],
        ]),

        "n_pool_kernel_size": trial.suggest_categorical("n_pool_kernel_size", [
            [8, 4, 2, 1], 
            [4, 2, 1, 1],
            [2, 1, 1, 1],
        ]),
        "scaler_type": "robust",
    }


auto_nhits = AutoNHITS(
    h=H,
    loss=MAE(),
    config=nhits_search_space,
    num_samples=20,
    backend="optuna",
    search_alg=optuna.samplers.TPESampler(seed=42),
    gpus=1,
    cpus=8,
    verbose=True,
)

nf_tuner = NeuralForecast(models=[auto_nhits], freq="5min", local_scaler_type='robust')
nf_cv = nf_tuner.cross_validation(df=data, val_size=8*VAL_SIZE, n_windows=1)

auto_nhits = nf_tuner.models[0]

trials_df = auto_nhits.results.trials_dataframe()

best_cfg = nf_tuner.models[0].results.best_trial.user_attrs["ALL_PARAMS"]
with open(os.path.join(SAVE_PATH, "best_cfg_192l.pkl"), "wb") as f:
    pickle.dump(best_cfg, f)
print("✅ Saved best hyperparameters to", SAVE_PATH)
```

```python
model = NHITS(
    # h=H,
    **best_cfg,
    precision="16-mixed",
    max_steps=1000,
    val_check_steps=50,
    early_stop_patience_steps=5,
    enable_checkpointing=True,
)

nf = NeuralForecast(models=[model], freq="5min")
nf.fit(df=data,val_size=VAL_SIZE)

ids   = data["unique_id"].unique()
dates = pd.date_range(start=FORECAST_DAY, periods=H, freq="5min")
future_df = pd.DataFrame({
    "unique_id": np.repeat(ids, len(dates)),
    "ds":         np.tile(dates, len(ids))
})
predictions = nf.predict(futr_df=future_df)
```

```python
def plot_forecast(location):

    # eval_date = datetime.datetime(2010,8,1)
    eval_date = FORECAST_DAY - datetime.timedelta(days=14)

    filtered_data = df_all[df_all['unique_id'] == location]
    filtered_data = filtered_data[(filtered_data['ds'] > eval_date) & (filtered_data['ds'] < datetime.datetime(2010,8,8))]
    filtered_nf_cv = nf_cv[nf_cv['unique_id'] == location]
    filtered_nf_cv = filtered_nf_cv[filtered_nf_cv['ds'] > eval_date]

    filtered_predictions = predictions[predictions['unique_id'] == location]

    cutoffs = filtered_nf_cv['cutoff'].unique()


    fig = go.Figure()
    fig.add_trace(go.Scatter(x=filtered_data['ds'], y=filtered_data['y'], mode='lines', name='Actual'))
    fig.add_trace(go.Scatter(x=filtered_nf_cv['ds'], y=filtered_nf_cv['AutoNHITS'], mode='lines', name='AutoNHITS', line=dict(dash='dash')))
    fig.add_trace(go.Scatter(x=filtered_predictions['ds'], y=filtered_predictions['NHITS'], mode='lines', name='AutoNHITS', line=dict(dash='dash')))

    for cutoff in cutoffs:
        fig.add_vline(x=cutoff, line=dict(color='black', dash='dot'))

    fig.update_layout(
        xaxis_title='Time steps',
        yaxis_title='Target [H1]',
        legend=dict(x=0, y=1),
        width=900,
        height=300
    )
    fig.show()

plot_forecast(42)
```
